```python
import pandas as pd

df = pd.read_csv("meteorología_2026.csv", skiprows=9)

df.head()
```

```python
datos = df.pivot_table(
    index=["date", "id_station"],
    columns="id_parameter",
    values="valor",
    aggfunc="first"
).reset_index()


# 3. Seleccionar estación
estacion = "CUT"

datos_estacion = datos[
    datos["id_station"] == estacion
].copy()


# 4. Seleccionar variables necesarias
datos_estacion = datos_estacion[
    ["date", "id_station", "RH", "WSP"]
].dropna()


# 5. Definir eventos
# H = Humedad alta
datos_estacion["humedad_alta"] = (
    datos_estacion["RH"] >= 70
)

# V = Viento tranquilo
datos_estacion["viento_tranquilo"] = (
    datos_estacion["WSP"] <= 1
)


# 6. Calcular probabilidades
# a) P(H)
p_humedad = datos_estacion[
    "humedad_alta"
].mean()

# b) P(V)
p_viento = datos_estacion[
    "viento_tranquilo"
].mean()

# c) P(H | V)
p_humedad_dado_viento = datos_estacion.loc[
    datos_estacion["viento_tranquilo"],
    "humedad_alta"
].mean()

# d) P(V | H)
p_viento_dado_humedad = datos_estacion.loc[
    datos_estacion["humedad_alta"],
    "viento_tranquilo"
].mean()


# 7. Mostrar resultados
print(f"Estación seleccionada: {estacion}")
print(f"Número de observaciones: {len(datos_estacion)}")
print()

print(
    f"a) P(Humedad alta) = "
    f"{p_humedad:.4f} = {p_humedad:.2%}"
)

print(
    f"b) P(Viento tranquilo) = "
    f"{p_viento:.4f} = {p_viento:.2%}"
)

print(
    f"c) P(Humedad alta | Viento tranquilo) = "
    f"{p_humedad_dado_viento:.4f} = "
    f"{p_humedad_dado_viento:.2%}"
)

print(
    f"d) P(Viento tranquilo | Humedad alta) = "
    f"{p_viento_dado_humedad:.4f} = "
    f"{p_viento_dado_humedad:.2%}"
)

```
